# Exploratory Data Analysis(EDA)-top_directors_by_reviews_cleaned

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
top_directors_by_reviews_cleaned = pd.read_csv("top_directors_by_reviews_cleaned.csv")
```

```python
numerical_cols = top_directors_by_reviews_cleaned.select_dtypes(include=['int64', 'float64']).columns
```

```python
# Summary statistics
print(top_directors_by_reviews_cleaned[numerical_cols].describe())
```

```python
# Histograms
for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(top_directors_by_reviews_cleaned[col].dropna(), kde=True, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

# Bivariate Analysis

```python
# Correlation matrix
plt.figure(figsize=(10, 6))
sns.heatmap(top_directors_by_reviews_cleaned[numerical_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

# Multivariate Analysis

```python
# Ensure no missing values
df_clean = top_directors_by_reviews_cleaned.dropna(subset=['director', 'director_avg_freshness', 'director_avg_reviews'])

plt.figure(figsize=(12, 6))
sns.scatterplot(
    data=df_clean,
    x='director',
    y='director_avg_freshness',
    hue='director_avg_reviews',
    #size='number_of_reviews',
    sizes=(20, 200),
    palette='viridis',
    alpha=0.5
)
plt.title("Number of Reviews vs Average Rating, Colored by Release Year")
plt.tight_layout()
plt.show()
```
